```python

from collections import deque

def snake_and_ladders():
    # Snakes: key = head, value = tail
    snakes = {
        99: 54,
        70: 55,
        52: 42,
        25: 2,
        95: 72
    }

    # Ladders: key = bottom, value = top
    ladders = {
        6: 25,
        11: 40,
        60: 85,
        46: 90,
        17: 69
    }

    # BFS queue stores (position, dice throws)
    queue = deque([(1, 0)])
    visited = set([1])

    while queue:
        position, throws = queue.popleft()

        # Destination reached
        if position == 100:
            print("Welcome to Snake and Ladders!")
            print("Starting position: 1")
            print("Destination: 100")
            print("Minimum dice throws required:", throws)
            return

        # Try all possible dice outcomes
        for dice in range(1, 7):
            next_position = position + dice

            if next_position > 100:
                continue

            # Apply ladder or snake
            if next_position in ladders:
                next_position = ladders[next_position]
            elif next_position in snakes:
                next_position = snakes[next_position]

            # Visit each position only once
            if next_position not in visited:
                visited.add(next_position)
                queue.append((next_position, throws + 1))

    print("Destination cannot be reached.")


snake_and_ladders()

```
